# Part C: Encoder swap + Shapley split (Shikshith V)

CLIP's image and text encoders meet only at a dot product, so features from two checkpoints can be mixed. With p = before and f = after, and letter 1 = vision checkpoint, letter 2 = text checkpoint, the four hybrids are **pp, pf, fp, ff**. Each is evaluated on the old domain (Oxford-IIIT Pets test set).

- F_V = 0.5 * ((A_pf - A_ff) + (A_pp - A_fp))
- F_L = 0.5 * ((A_fp - A_ff) + (A_pp - A_pf))
- F_V + F_L = A_pp - A_ff (total forgetting)
